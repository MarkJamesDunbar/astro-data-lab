```python
import lightkurve as lk
from pathlib import Path
import pandas as pd

```

```python
DATA_ROOT = Path("/data/astro-data-lab")
TESS_DIR = DATA_ROOT / "raw" / "tess"
LABEL_DIR = DATA_ROOT / "raw" / "labels"

TESS_DIR.mkdir(parents=True, exist_ok=True)
LABEL_DIR.mkdir(parents=True, exist_ok=True)
```

# TESS Data

```python
target = "TIC 307210830"

search_results = lk.search_lightcurve(
    target,
    mission="TESS"
)

search_results
```

```python
lc = search_results[0].download(
    download_dir=str(TESS_DIR)
)
lc
```

```python
lc.to_pandas().head()
```

```python
lc.plot()
```

```python

from astroquery.mast import Observations

# Count all TESS time-series observations
total = Observations.query_criteria_count(
    obs_collection="TESS",
    dataproduct_type="timeseries"
)

print(f"Total TESS observations: {total:,}")

# Inspect Sector 1
observations = Observations.query_criteria(
    obs_collection="TESS",
    dataproduct_type="timeseries",
    sequence_number=1
)

print(f"Sector 1: {len(observations):,} observations")
display(observations[:5].to_pandas())
```

# Labels

```python
LABEL_URL = (
    "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?"
    "query=select+*+from+ps+where+default_flag=1"
    "&format=csv"
)

planets = pd.read_csv(LABEL_URL)

planets.head()
```

```python
label_path = LABEL_DIR / "confirmed_planets.csv"

planets.to_csv(label_path, index=False)

print(f"Saved {len(planets):,} rows to {label_path}")
```
